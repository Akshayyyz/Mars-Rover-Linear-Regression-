```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('titanic (1).csv')
data = df.copy()

data = data.drop(columns=['PassengerId', 'Name', 'Ticket', 'Cabin'])

data['Age'] = data['Age'].fillna(data['Age'].median())
data['Embarked'] = data['Embarked'].fillna(data['Embarked'].mode()[0])

data['Sex'] = data['Sex'].map({'male': 0, 'female': 1})
data = pd.get_dummies(data, columns=['Embarked'], drop_first=True, dtype=float)

feature_names = list(data.drop(columns=['Survived']).columns)
X = data.drop(columns=['Survived']).values
y = data['Survived'].values

indices = np.random.permutation(len(X))
split_idx = int(0.8 * len(X))
train_idx, test_idx = indices[:split_idx], indices[split_idx:]

X_train_raw, X_test_raw = X[train_idx], X[test_idx]
y_train, y_test = y[train_idx], y[test_idx]

mean = np.mean(X_train_raw, axis=0)
std = np.std(X_train_raw, axis=0)
X_train = (X_train_raw - mean) / (std + 1e-8)
X_test = (X_test_raw - mean) / (std + 1e-8)
```

```python
class LogisticRegressionScratch:
    def __init__(self, learning_rate=0.1, epochs=1000, reg_lambda=0.01):
        self.lr = learning_rate
        self.epochs = epochs
        self.reg_lambda = reg_lambda
        self.weights = None
        self.bias = 0.0
        self.train_losses = []
        self.val_losses = []

    def _sigmoid(self, z):
        z = np.clip(z, -250, 250)  # Prevent overflow
        return 1.0 / (1.0 + np.exp(-z))

    def compute_loss(self, y_true, y_prob):
        m = len(y_true)
        eps = 1e-15
        y_prob = np.clip(y_prob, eps, 1 - eps)
        bce = - (1.0 / m) * np.sum(y_true * np.log(y_prob) + (1 - y_true) * np.log(1 - y_prob))
        l2_penalty = (self.reg_lambda / (2.0 * m)) * np.sum(self.weights ** 2)
        return bce + l2_penalty

    def fit(self, X, y, X_val=None, y_val=None):
        m, n = X.shape
        self.weights = np.zeros(n)
        self.bias = 0.0

        for epoch in range(self.epochs):
            linear_out = np.dot(X, self.weights) + self.bias
            y_prob = self._sigmoid(linear_out)

            dw = (1.0 / m) * np.dot(X.T, (y_prob - y)) + (self.reg_lambda / m) * self.weights
            db = (1.0 / m) * np.sum(y_prob - y)

            self.weights -= self.lr * dw
            self.bias -= self.lr * db

            self.train_losses.append(self.compute_loss(y, y_prob))
            if X_val is not None and y_val is not None:
                val_prob = self.predict_proba(X_val)
                self.val_losses.append(self.compute_loss(y_val, val_prob))

    def predict_proba(self, X):
        return self._sigmoid(np.dot(X, self.weights) + self.bias)

    def predict(self, X, threshold=0.5):
        return (self.predict_proba(X) >= threshold).astype(int)
```

```python
model = LogisticRegressionScratch(learning_rate=0.1, epochs=1000, reg_lambda=0.01)
model.fit(X_train, y_train, X_val=X_test, y_val=y_test)

y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)

TP = np.sum((y_test == 1) & (y_pred == 1))
TN = np.sum((y_test == 0) & (y_pred == 0))
FP = np.sum((y_test == 0) & (y_pred == 1))
FN = np.sum((y_test == 1) & (y_pred == 0))

accuracy = (TP + TN) / len(y_test)
precision = TP / (TP + FP) if (TP + FP) > 0 else 0.0
recall = TP / (TP + FN) if (TP + FN) > 0 else 0.0
f1_score = (2 * precision * recall) / (precision + recall) if (precision + recall) > 0 else 0.0

thresholds = np.linspace(0, 1, 200)
tpr_list, fpr_list = [], []
for thresh in thresholds:
    preds = (y_prob >= thresh).astype(int)
    tp = np.sum((y_test == 1) & (preds == 1))
    tn = np.sum((y_test == 0) & (preds == 0))
    fp = np.sum((y_test == 0) & (preds == 1))
    fn = np.sum((y_test == 1) & (preds == 0))
    tpr_list.append(tp / (tp + fn) if (tp + fn) > 0 else 0.0)
    fpr_list.append(fp / (fp + tn) if (fp + tn) > 0 else 0.0)

sorted_idx = np.argsort(fpr_list)
fpr_sorted = np.array(fpr_list)[sorted_idx]
tpr_sorted = np.array(tpr_list)[sorted_idx]
roc_auc = np.trapezoid(tpr_sorted, fpr_sorted)

print(f"Accuracy : {accuracy:.4f} ({accuracy*100:.2f}%)")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1-Score : {f1_score:.4f}")
print(f"ROC-AUC  : {roc_auc:.4f}")
```

```python
%matplotlib inline
plt.style.use('seaborn-v0_8-whitegrid')
fig2, axes2 = plt.subplots(2, 2, figsize=(14, 10))
fig2.suptitle('Logistic Regression From Scratch — Training & Evaluation Diagnostics', fontsize=16, fontweight='bold')

axes2[0, 0].plot(model.train_losses, label='Train BCE Loss', color='#2980b9', linewidth=2)
axes2[0, 0].plot(model.val_losses, label='Validation BCE Loss', color='#e67e22', linewidth=2, linestyle='--')
axes2[0, 0].set_title('Binary Cross-Entropy Loss Curve Over Epochs', fontweight='bold')
axes2[0, 0].set_xlabel('Epochs')
axes2[0, 0].set_ylabel('Loss')
axes2[0, 0].legend()

axes2[1, 0].plot(fpr_sorted, tpr_sorted, color='#8e44ad', linewidth=2.5, label=f'Logistic Regression (AUC = {roc_auc:.4f})')
axes2[1, 0].plot([0, 1], [0, 1], color='gray', linestyle='--', label='Random Guess (AUC = 0.50)')
axes2[1, 0].fill_between(fpr_sorted, tpr_sorted, alpha=0.15, color='#8e44ad')
axes2[1, 0].set_title('Receiver Operating Characteristic (ROC) Curve', fontweight='bold')
axes2[1, 0].set_xlabel('False Positive Rate (FPR)')
axes2[1, 0].set_ylabel('True Positive Rate (TPR / Recall)')
axes2[1, 0].legend(loc='lower right')

sorted_w_idx = np.argsort(model.weights)
sorted_features = [feature_names[i] for i in sorted_w_idx]
sorted_weights = model.weights[sorted_w_idx]
colors = ['#e74c3c' if w < 0 else '#2ecc71' for w in sorted_weights]
axes2[1, 1].barh(sorted_features, sorted_weights, color=colors, edgecolor='black')
axes2[1, 1].axvline(0, color='black', linewidth=1)
axes2[1, 1].set_title('Learned Model Weights (Feature Importance)', fontweight='bold')
axes2[1, 1].set_xlabel('Standardized Weight Value (w)')

plt.tight_layout()
plt.show()
```
